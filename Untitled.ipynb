```python
import glob, os
import pandas as pd

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", None)

dfs = {}
for f in sorted(glob.glob("*.csv")):
    name = os.path.splitext(f)[0]
    dfs[name] = pd.read_csv(f)

    print("=" * 60)
    print(name.upper())
    print("=" * 60)
    dfs[name].info()
    print(dfs[name].head())
    print()

for name, df in dfs.items():
    date_cols = [c for c in df.columns if "date" in c.lower()]
    for c in date_cols:
        missing_before = df[c].isna().sum()
        df[c] = pd.to_datetime(df[c], errors="coerce")
        failed = df[c].isna().sum() - missing_before
        print(f"{name}.{c}: converted, {failed} values could not be read")

dfs["marketing_spend"]["month"] = pd.to_datetime(dfs["marketing_spend"]["month"], errors="coerce")
```

```python
o = dfs['orders'].copy()
c = dfs['customers'].copy()
oc = o.loc[o['order_status'] == 'completed']

no_completed = c[~c['customer_id'].isin(oc['customer_id'])]
print(no_completed.shape)      # (3565, 7)
no_completed.head()
```

```python
first = oc.groupby('customer_id')['order_date'].min().rename('first_order_date')
c = c.merge(first, on='customer_id', how='left')   # left join keeps non-buyers
c['bought'] = c['first_order_date'].notna()
c['days_to_first'] = (c['first_order_date'] - c['signup_date']).dt.days
```
